1) In quantum mechanics, the Hamiltonian is the operator that represents the total energy of a system. It contains both the kinetic energy and the potential energy of the particle. The Hamiltonian is extremely important because solving the Schrödinger equation with the Hamiltonian gives the allowed energy levels and wave functions of the system. Sources: https://en.wikipedia.org/wiki/Hamiltonian_(quantum_mechanics), https://physics.stackexchange.com/questions/488218/what-is-a-hamiltonian-of-a-system
2) The function represents Hooke’s Law, which describes the restoring force acting on a simple harmonic oscillator.
3) The eigenstates are the permitted stationary states of the quantum system. Solving the Schrödinger equation with the Hamiltonian yields discrete energy values, known as eigenvalues, along with their associated wave functions, called eigenstates. Every eigenstate corresponds to a specific, well-defined energy. In the case of the harmonic oscillator, these eigenstates represent quantized vibrational modes. Eigenstates are important because a measurement of the particle’s energy while it is in one of these states will always produce the same energy value. Sources: https://en.wikipedia.org/wiki/Eigenvalues_and_eigenvectors, https://phys.libretexts.org/Bookshelves/Quantum_Mechanics/Introductory_Quantum_Mechanics_(Fitzpatrick)/03%3A_Fundamentals_of_Quantum_Mechanics/3.08%3A_Eigenstates_and_Eigenvalues

```python
import numpy as np
import matplotlib.pyplot as plt

from qmsolve import Hamiltonian, SingleParticle, init_visualization, Å, eV
```

```python
# Harmonic oscillator potential
def harmonic_oscillator(particle):
    k = 100 * eV / Å**2
    return 0.5 * k * particle.x**2

# Define the Hamiltonian
H = Hamiltonian(particles = SingleParticle(), potential = harmonic_oscillator, 
                spatial_ndim = 1, N = 512, extent = 20*Å)

#Diagonalize the Hamiltonian and compute the eigenstates
eigenstates = H.solve(max_states = 30)

# Print out the energies of the eigenstates
print(eigenstates.energies) # the printed energies are expressed in eV

# Visualize the Eigenstates
visualization = init_visualization(eigenstates)
visualization.plot_eigenstate(0)
```

### New potential: infinite square well

Below I define the infinite square well potential for $L = 8$ Angstroms and plot the two lowest-energy eigenstates (compare to Figure 6.3 in your textbook).

Answers:
5) The x-axis represents the position of the particle, in Angstroms (Å). Because the Hamiltonian is defined in one spatial dimension and the potential depends on the particle’s position. The extent = 20*Å parameter sets the spatial range over which the wave functions are calculated.
6) The plots show the wave functions ψ(x), not the probability densities. The plotted curves contain both positive and negative values. A probability density,∣ψ(x)∣^2 can never be negative since it is the square magnitude of the wave function.

```python
# 8 angstrom-wide infinite square well
def infinite_square_well(particle):
    potential = np.zeros_like(particle.x)
    out_of_bounds = np.abs(particle.x) > 4*Å
    potential[out_of_bounds] = np.inf
    return potential

# Define the Hamiltonian of the infinite square well
H_sw = Hamiltonian(particles = SingleParticle(), potential = infinite_square_well, 
                   spatial_ndim = 1, N = 512, extent = 10*Å)

# Diagonalize the Hamiltonian and compute the eigenstates
eigenstates_sw = H_sw.solve(max_states = 30)

print(eigenstates_sw.energies)

# Visualize the infinite square well eigenstates
visualization_sw = init_visualization(eigenstates_sw)
visualization_sw.plot_eigenstate(0)
visualization_sw.plot_eigenstate(1)
```

Answers: 
8) The eigenstates of the infinite square well are exactly zero at the boundaries of the well. This occurs because the potential outside the box is infinite, meaning the particle cannot exist beyond those limits. As a result, the wave function must obey the boundary conditions ψ(−L/2)=0 and ψ(L/2)=0. The simple harmonic oscillator (SHO) eigenstates differ because the harmonic oscillator potential remains finite everywhere. Instead of abruptly becoming zero, the SHO wave functions extend infinitely in space and gradually approach zero asymptotically. Another key distinction is that the infinite square well wave functions are completely confined within a finite interval, whereas the SHO wave functions spread throughout all space.
9) As the quantum number increases, the number of extrema also increases. In the infinite square well, the n=1 state contains 1 extremum, the n=2 state contains 2 extrema, and the n=3 state contains 3 extrema. More generally, the number of extrema is equal to n. States with higher energy oscillate more rapidly, which leads to a greater number of maxima and minima in the wave function.

```python
def finite_square_well(particle):
    V0 = 20 * eV
    width = 8 * Å

    potential = np.full_like(particle.x, V0)

    inside = np.abs(particle.x) < width/2
    potential[inside] = 0

    return potential

H_finite = Hamiltonian(
    particles = SingleParticle(),
    potential = finite_square_well,
    spatial_ndim = 1,
    N = 512,
    extent = 20*Å
)

finite_states = H_finite.solve(max_states = 30)


visualization_finite = init_visualization(finite_states)

visualization_finite.plot_eigenstate(1)
visualization_finite.plot_eigenstate(9)
visualization_finite.plot_eigenstate(16)
visualization_finite.plot_eigenstate(20)
visualization_finite.plot_eigenstate(12)
```
